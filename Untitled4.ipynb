```python
# ============================================================
# GYM FORM CORRECTOR - ALL EXERCISES WITH STRICT FORM CHECKS
# ============================================================

import cv2
import numpy as np
import pandas as pd
import base64
import time

from ultralytics import YOLO
from IPython.display import display, Javascript, Image
from google.colab.output import eval_js

print("Libraries imported successfully!")

# Load pretrained YOLO Pose model
model = YOLO("yolo11n-pose.pt")
print("YOLO Pose model loaded successfully!\n")


# ============================================================
# 1. ANGLE CALCULATION & HELPER FUNCTIONS
# ============================================================

def calculate_angle(a, b, c):
    """Calculates 2D angle (in degrees) between three points with b as vertex."""
    a, b, c = np.array(a), np.array(b), np.array(c)
    radians = np.arctan2(c[1] - b[1], c[0] - b[0]) - np.arctan2(a[1] - b[1], a[0] - b[0])
    angle = np.abs(radians * 180.0 / np.pi)
    return 360.0 - angle if angle > 180.0 else angle


def get_best_side(kp, conf, left_idxs, right_idxs, min_conf=0.45):
    """Dynamically picks the left or right side based on detection confidence."""
    left_conf = np.mean([conf[i] for i in left_idxs])
    right_conf = np.mean([conf[i] for i in right_idxs])

    if left_conf >= right_conf:
        chosen_idxs, avg_conf = left_idxs, left_conf
    else:
        chosen_idxs, avg_conf = right_idxs, right_conf

    if avg_conf < min_conf:
        return None, avg_conf

    return [kp[i] for i in chosen_idxs], avg_conf


# ============================================================
# 2. STRICT FORM CHECKERS FOR ALL EXERCISES
# ============================================================

def check_bicep_curl(kp, conf):
    # Require clear keypoints for BOTH shoulders, elbows, wrists, and hips
    req_indices = [5, 6, 7, 8, 9, 10, 11, 12]
    if any(conf[i] < 0.45 for i in req_indices):
        return 0, "WRONG", "Step back: Both arms & hips must be clearly visible"

    l_shoulder, r_shoulder = kp[5], kp[6]
    l_elbow, r_elbow = kp[7], kp[8]
    l_wrist, r_wrist = kp[9], kp[10]
    l_hip, r_hip = kp[11], kp[12]

    # Calculate angles for both left and right arms independently
    l_angle = calculate_angle(l_shoulder, l_elbow, l_wrist)
    r_angle = calculate_angle(r_shoulder, r_elbow, r_wrist)

    # Pick the arm that is actually flexing (smaller elbow angle)
    if l_angle < r_angle:
        shoulder, elbow, wrist, hip = l_shoulder, l_elbow, l_wrist, l_hip
        angle = l_angle
        opp_wrist = r_wrist
    else:
        shoulder, elbow, wrist, hip = r_shoulder, r_elbow, r_wrist, r_hip
        angle = r_angle
        opp_wrist = l_wrist

    # 1. FALSE-POSITIVE CHECK: Hand near face/chin/mouth (Touching face is NOT a curl)
    nose = kp[0] if conf[0] > 0.45 else shoulder
    if abs(wrist[1] - nose[1]) < 40 or wrist[1] < shoulder[1] - 20:
        return angle, "WRONG", "Invalid form: Keep hand away from face/chin"

    # 2. FALSE-POSITIVE CHECK: Hand crossing over chest centerline
    chest_x = (l_shoulder[0] + r_shoulder[0]) / 2
    if abs(wrist[0] - chest_x) < 30:
        return angle, "WRONG", "Keep weight out in front of shoulder, not on chest"

    # 3. ELBOW STABILITY CHECK: Elbow must remain pinned near ribs
    if abs(elbow[0] - shoulder[0]) > 50:
        return angle, "WRONG", "Keep elbow pinned close to your torso"

    # 4. POSTURE CHECK: Torso must be vertical
    if abs(hip[1] - shoulder[1]) < 60:
        return angle, "WRONG", "Stand up straight"

    # Final Curl Depth Evaluation
    if angle < 55:
        return angle, "CORRECT", "Good curl contraction"
    elif angle > 150:
        return angle, "WRONG", "Arm extended: Curl upward"
    else:
        return angle, "WRONG", "Incomplete range: Curl higher"


def check_squat(kp, conf):
    pts, avg_conf = get_best_side(kp, conf, [11, 13, 15], [12, 14, 16])
    shoulder_pts, s_conf = get_best_side(kp, conf, [5], [6])

    # Visibility Check
    if pts is None or shoulder_pts is None or s_conf < 0.45:
        return 0, "WRONG", "Step back: Full body (head to feet) required"

    shoulder = shoulder_pts[0]
    hip, knee, ankle = pts

    # Posture Check (Torso must stay upright)
    if abs(shoulder[0] - hip[0]) > 90:
        return 0, "WRONG", "Back leaning too far forward: Keep chest up"

    angle = calculate_angle(hip, knee, ankle)

    if angle < 100:
        # Knee Caving/Over-extension Check
        if knee[0] > ankle[0] + 45:
            return angle, "WRONG", "Knees over toes: Shift weight to heels"
        return angle, "CORRECT", "Good squat depth"
    elif angle > 160:
        return angle, "WRONG", "Standing stance: Lower down into squat"
    else:
        return angle, "WRONG", "Incomplete depth: Go lower"


def check_pushup(kp, conf):
    arm_pts, arm_conf = get_best_side(kp, conf, [5, 7, 9], [6, 8, 10])
    body_pts, body_conf = get_best_side(kp, conf, [5, 11, 15], [6, 12, 16])

    # Visibility Check
    if arm_pts is None or body_pts is None or arm_conf < 0.45 or body_conf < 0.45:
        return 0, "WRONG", "Step back: Full horizontal body required"

    shoulder, hip, ankle = body_pts
    elbow_angle = calculate_angle(arm_pts[0], arm_pts[1], arm_pts[2])
    body_angle = calculate_angle(shoulder, hip, ankle)

    # Posture Check (Prevent sagging or raised hips)
    if body_angle < 155:
        return elbow_angle, "WRONG", "Keep back and hips in a straight line"

    # Horizontal Plank Check
    if abs(shoulder[1] - ankle[1]) < 30:
        return elbow_angle, "WRONG", "Must be on the floor in a plank position"

    if elbow_angle < 100:
        return elbow_angle, "CORRECT", "Good push-up depth"
    elif elbow_angle > 160:
        return elbow_angle, "WRONG", "Top lockout: Lower chest down"
    else:
        return elbow_angle, "WRONG", "Lower chest further down"


def check_lunge(kp, conf):
    pts, avg_conf = get_best_side(kp, conf, [11, 13, 15], [12, 14, 16])
    shoulder_pts, s_conf = get_best_side(kp, conf, [5], [6])

    # Visibility Check
    if pts is None or shoulder_pts is None or s_conf < 0.45:
        return 0, "WRONG", "Step back: Full body must be visible"

    shoulder = shoulder_pts[0]
    hip, knee, ankle = pts

    # Torso Orientation Check
    if abs(shoulder[0] - hip[0]) > 60:
        return 0, "WRONG", "Torso leaning: Keep upper body upright"

    angle = calculate_angle(hip, knee, ankle)

    if 70 <= angle <= 105:
        if knee[0] > ankle[0] + 40:
            return angle, "WRONG", "Front knee too far forward over toe"
        return angle, "CORRECT", "Good lunge angle"
    elif angle > 155:
        return angle, "WRONG", "Standing: Step forward & bend knee"
    else:
        return angle, "WRONG", "Adjust knee bend to 90 degrees"


def check_shoulder_press(kp, conf):
    pts, avg_conf = get_best_side(kp, conf, [5, 7, 9], [6, 8, 10])
    hip_pts, hip_conf = get_best_side(kp, conf, [11], [12])

    # Visibility Check
    if pts is None or hip_pts is None or hip_conf < 0.45:
        return 0, "WRONG", "Step back: Upper body & hips must be visible"

    shoulder, elbow, wrist = pts
    hip = hip_pts[0]

    # Posture Check (Prevent leaning back excessively)
    if abs(hip[0] - shoulder[0]) > 50:
        return 0, "WRONG", "Do not lean back: Stand upright"

    angle = calculate_angle(shoulder, elbow, wrist)

    if angle > 155:
        # Check if wrist extends above shoulder height
        if wrist[1] > shoulder[1]:
            return angle, "WRONG", "Hands must extend above head"
        return angle, "CORRECT", "Good overhead press extension"
    elif angle < 90:
        return angle, "WRONG", "Bottom position: Press weights upward"
    else:
        return angle, "WRONG", "Lock out arms fully overhead"


def check_plank(kp, conf):
    pts, avg_conf = get_best_side(kp, conf, [5, 11, 15], [6, 12, 16])

    # Visibility Check
    if pts is None or avg_conf < 0.45:
        return 0, "WRONG", "Step back: Full horizontal body required"

    shoulder, hip, ankle = pts

    # Horizontal Position Check
    if abs(shoulder[1] - ankle[1]) < 30:
        return 0, "WRONG", "Get down on the floor to perform plank"

    angle = calculate_angle(shoulder, hip, ankle)

    if angle >= 165:
        return angle, "CORRECT", "Good straight plank alignment"
    elif angle < 150:
        return angle, "WRONG", "Hips sagging: Lift core up"
    else:
        return angle, "WRONG", "Keep shoulders, hips & feet aligned"


def check_jumping_jacks(kp, conf):
    # Front-facing Full Body Check
    required_indices = [5, 6, 9, 10, 11, 12, 15, 16]
    if any(conf[i] < 0.45 for i in required_indices):
        return 0, "WRONG", "Step back: Entire front body must face camera"

    left_wrist, right_wrist = kp[9], kp[10]
    left_ankle, right_ankle = kp[15], kp[16]
    left_shoulder, right_shoulder = kp[5], kp[6]

    hand_dist = abs(left_wrist[0] - right_wrist[0])
    feet_dist = abs(left_ankle[0] - right_ankle[0])

    # Validate arms are elevated above shoulder height
    hands_raised = left_wrist[1] < left_shoulder[1] and right_wrist[1] < right_shoulder[1]

    if hand_dist > 160 and feet_dist > 110 and hands_raised:
        return feet_dist, "CORRECT", "Good jump extension"
    elif hand_dist < 80 and feet_dist < 60:
        return feet_dist, "WRONG", "Neutral stance: Jump arms and feet out"
    else:
        return feet_dist, "WRONG", "Raise hands fully overhead and jump wide"


def check_exercise(exercise, kp, conf):
    checkers = {
        "Squat": check_squat,
        "Bicep Curl": check_bicep_curl,
        "Push-up": check_pushup,
        "Lunge": check_lunge,
        "Shoulder Press": check_shoulder_press,
        "Plank": check_plank,
        "Jumping Jacks": check_jumping_jacks,
    }
    checker = checkers.get(exercise)
    return checker(kp, conf) if checker else (0, "WRONG", "Unknown exercise")


# ============================================================
# 3. EXERCISE SELECTION
# ============================================================

print("Choose your exercise:")
print("1 - Squat")
print("2 - Bicep Curl")
print("3 - Push-up")
print("4 - Lunge")
print("5 - Shoulder Press")
print("6 - Plank")
print("7 - Jumping Jacks\n")

choice = input("Enter number (1-7): ").strip()

exercise_list = {
    "1": "Squat",
    "2": "Bicep Curl",
    "3": "Push-up",
    "4": "Lunge",
    "5": "Shoulder Press",
    "6": "Plank",
    "7": "Jumping Jacks",
}

exercise = exercise_list.get(choice, "Squat")
print(f"\nSelected Exercise: {exercise}\n")


# ============================================================
# 4. TRACKING COUNTERS
# ============================================================

counter = 0
stage = None
correct_frames = 0
wrong_frames = 0


# ============================================================
# 5. CAMERA JAVASCRIPT ENGINE
# ============================================================

display(
    Javascript("""
window.gymCameraRunning = false;

window.startGymCamera = async function() {
    const old = document.getElementById('gym-camera-container');
    if (old) { old.remove(); }

    const container = document.createElement('div');
    container.id = 'gym-camera-container';

    const video = document.createElement('video');
    video.id = 'gym-video';
    video.width = 640;
    video.height = 480;
    video.style.border = '3px solid green';

    const button = document.createElement('button');
    button.innerHTML = 'STOP CAMERA';
    button.style.fontSize = '18px';
    button.style.padding = '10px';
    button.style.margin = '10px';

    container.appendChild(video);
    container.appendChild(button);
    document.body.appendChild(container);

    const stream = await navigator.mediaDevices.getUserMedia({
        video: { width: 640, height: 480 },
        audio: false
    });

    video.srcObject = stream;
    await video.play();

    window.gymCameraRunning = true;

    button.onclick = function() {
        window.gymCameraRunning = false;
        stream.getTracks().forEach(track => track.stop());
        container.remove();
    };

    return true;
};

window.getGymFrame = function() {
    const video = document.getElementById('gym-video');
    if (!video) return null;

    const canvas = document.createElement('canvas');
    canvas.width = video.videoWidth;
    canvas.height = video.videoHeight;

    const ctx = canvas.getContext('2d');
    ctx.drawImage(video, 0, 0, canvas.width, canvas.height);

    return canvas.toDataURL('image/jpeg', 0.7);
};

window.isGymCameraRunning = function() {
    return window.gymCameraRunning;
};
""")
)

eval_js("startGymCamera()")
print("Camera started. Step back so your full posture is visible.\n")


# ============================================================
# 6. LIVE WEBCAM PROCESSING LOOP
# ============================================================

display_handle = display(Image(data=b""), display_id=True)

while True:
    running = eval_js("isGymCameraRunning()")
    if not running:
        break

    data = eval_js("getGymFrame()")
    if data is None:
        continue

    binary = base64.b64decode(data.split(",")[1])
    frame = cv2.imdecode(np.frombuffer(binary, np.uint8), cv2.IMREAD_COLOR)

    results = model(frame, verbose=False)
    annotated_frame = frame.copy()

    if (
        len(results) > 0
        and results[0].keypoints is not None
        and len(results[0].keypoints.data) > 0
    ):
        kp = results[0].keypoints.xy.cpu().numpy()[0]
        conf = results[0].keypoints.conf.cpu().numpy()[0]

        if len(kp) > 0 and len(conf) > 0:
            annotated_frame = results[0].plot()

            angle, status, feedback = check_exercise(exercise, kp, conf)

            if status == "CORRECT":
                correct_frames += 1
            else:
                wrong_frames += 1

            # Repetition tracking logic
            if exercise in ["Squat", "Push-up", "Lunge"]:
                if angle < 100 and status == "CORRECT":
                    stage = "down"
                if angle > 155 and stage == "down":
                    stage = "up"
                    counter += 1

            elif exercise == "Bicep Curl":
                if angle < 50 and status == "CORRECT":
                    stage = "up"
                if angle > 150 and stage == "up":
                    stage = "down"
                    counter += 1

            elif exercise == "Shoulder Press":
                if angle < 90:
                    stage = "down"
                if angle > 155 and stage == "down" and status == "CORRECT":
                    stage = "up"
                    counter += 1

            elif exercise == "Jumping Jacks":
                if angle > 110 and status == "CORRECT":
                    stage = "open"
                if angle < 60 and stage == "open":
                    stage = "closed"
                    counter += 1

            # Visual overlay
            banner_color = (0, 200, 0) if status == "CORRECT" else (0, 0, 220)
            text_str = (
                "FORM IS CORRECT" if status == "CORRECT" else "FORM IS WRONG"
            )

            cv2.rectangle(annotated_frame, (0, 0), (640, 50), banner_color, -1)
            cv2.putText(
                annotated_frame,
                text_str,
                (180, 35),
                cv2.FONT_HERSHEY_SIMPLEX,
                1.0,
                (255, 255, 255),
                3,
            )

            cv2.putText(
                annotated_frame,
                f"Exercise: {exercise}",
                (20, 85),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                (255, 255, 255),
                2,
            )
            cv2.putText(
                annotated_frame,
                f"Metric: {int(angle)}",
                (20, 115),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                (255, 255, 255),
                2,
            )

            if exercise != "Plank":
                cv2.putText(
                    annotated_frame,
                    f"Reps: {counter}",
                    (20, 145),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.65,
                    (255, 255, 255),
                    2,
                )

            cv2.putText(
                annotated_frame,
                f"Feedback: {feedback}",
                (20, 180),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                (0, 255, 255),
                2,
            )

    else:
        cv2.rectangle(annotated_frame, (0, 0), (640, 50), (0, 0, 220), -1)
        cv2.putText(
            annotated_frame,
            "NO PERSON DETECTED",
            (170, 35),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.9,
            (255, 255, 255),
            2,
        )

    annotated_frame = cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB)
    success, buffer = cv2.imencode(".jpg", annotated_frame)

    if success:
        display_handle.update(Image(data=buffer.tobytes()))

    time.sleep(0.03)


# ============================================================
# 7. PRINT SUMMARY AFTER CAMERA STOPS
# ============================================================

total_frames = correct_frames + wrong_frames
accuracy = (correct_frames / total_frames * 100) if total_frames > 0 else 0

print("Camera stopped.\n")
print("=" * 45)
print("             SESSION SUMMARY")
print("=" * 45)
print(f"Exercise Selected      : {exercise}")
print(f"Total Repetitions      : {counter}")
print(f"Correct Pose Frames    : {correct_frames}")
print(f"Wrong Pose Frames      : {wrong_frames}")
print(f"Overall Form Accuracy  : {accuracy:.1f}%")
print("=" * 45)

workout_plan = {
    "Monday": {
        "Focus": "Chest + Triceps",
        "Exercises": [
            ("Push-up", "3 × 12"),
            ("Shoulder Press", "3 × 10"),
            ("Plank", "3 × 30 sec"),
        ],
    },
    "Tuesday": {
        "Focus": "Legs & Cardio",
        "Exercises": [
            ("Squat", "3 × 12"),
            ("Lunge", "3 × 10 each leg"),
            ("Jumping Jacks", "3 × 45 sec"),
        ],
    },
    "Wednesday": {
        "Focus": "Biceps + Core",
        "Exercises": [("Bicep Curl", "3 × 12"), ("Plank", "3 × 30 sec")],
    },
    "Thursday": {
        "Focus": "Recovery",
        "Exercises": [("Walking", "20–30 min"), ("Stretching", "10 min")],
    },
    "Friday": {
        "Focus": "Full Body HIIT",
        "Exercises": [
            ("Jumping Jacks", "3 × 45 sec"),
            ("Squat", "3 × 12"),
            ("Push-up", "3 × 10"),
            ("Bicep Curl", "3 × 12"),
        ],
    },
    "Saturday": {
        "Focus": "Legs + Core",
        "Exercises": [
            ("Squat", "3 × 15"),
            ("Lunge", "3 × 10"),
            ("Plank", "3 × 30 sec"),
        ],
    },
    "Sunday": {"Focus": "Rest", "Exercises": [("Recovery", "Rest")]},
}

print("\n" + "=" * 45)
print("             WEEKLY WORKOUT PLAN")
print("=" * 45)

for day, plan in workout_plan.items():
    print(f"\n{day}")
    print(f"Focus: {plan['Focus']}")
    print("-" * 45)
    for exercise_name, reps in plan["Exercises"]:
        print(f"  {exercise_name} -> {reps}")
```
